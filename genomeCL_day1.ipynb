## Navigating the command line
This workshop assumes you have some familiarity with Unix file systems, but let's do a quick review for how we orient ourselves and move around in a file system. We are working in a `jupyter` notebook in this workshop, so the file system layout will be different from when you are at your own terminal or logged onto the Harvard `cannon` HPC cluster, but the general principles should be the same.

A file system is a hierarchical tree of directories (aka "folders"), with files stored within those directories. The "highest" level directory in a Unix-like (e.g. Mac and Linux) system is the `root` directory (specified by `/`); all other directories are stored recursively within the root directory and are "lower" in the tree. We can check our position within the tree using the Unix command `pwd` (which stands for `p`resent `w`orking `d`irectory):

````python
%%bash
pwd
````

This prints our current path, i.e. a list of all directories separated by a `/` starting from the root on down. You can think of a path as an "address" where something "lives" in a file system. To display the contents of our current directory, we use the `ls` command:

````python
%%bash
ls
````

This prints out a list of files in the directory we are currently in, as well as any subdirectories, i.e. directories that are nested further down in the tree. As we did not specify any after our `ls` command, it listed the content of our current directory, but we can also give a path to `ls` and it will list the contents of the directory at that path, e.g.

````python
%%bash
ls data_day1/
````

In this example we are specifying a *relative path*, meaning relative to where we currently are in the tree; in other words, the `ls` commands looks for a directory called `data_day1/` in our current directory. We can instead use an *absolute path* to a directory or file, which list the complete address starting from the root:

````python
%%bash
ls /content/data_day1/
````

If we want to change our current directory, we use the `cd` (`c`hange `d`irectory) command followed by a path:

```
cd /directory/to/move/to
```

The path can be an absolute path, or a relative path. If it is a relative path, `cd` will look for the directory to move to in your current directory, and will produce an error if it cannot find it. In general, always be mindful of whether you are specifying absolute or relative path when running commands, as it is easy to use the wrong path and cause a pipeline to fail when it cannot find a file!

> Quick tip!
> Some very common command line shortcuts you will probably run into are `.` and `..`. The `.` means "the directory I am currently in", and `..` = "the directory one level above my current directory".
> So executing `cd ..` or `ls ..` would change my directory to the one above or list the contents of the directory above (respectively).
> You can even chain them together, so `ls ../../` would list the contents *two* directories above, etc...

#### Using arguments
So far we have run the different commands using their default functions. However, almost all programs will have optional **arguments** (also called options, parameters, flags) that modify their default behavior in some way. They are specified after the main command and are usually (but not always!) denoted with a `-` or `--`. Multiple arguments can be provided at once, though depending on the program certain arguments may be mutually exclusive with each other or have other certain requirements. For example, we can modify how `ls` lists a directory contents by using `-l` to output the "long list" format, which includes extra information such as file owner, permissions, and file size in bytes.

````python
%%bash
ls -l
````

We can pair `-l` with with the `-h` argument which displays the file size in a convenient "human-readable" format (rather than the raw number of bytes). This is a case where one argument requires another, as on its own `-h` won't display anything (as by default `ls` doesn't output file size):

````python
%%bash
ls -l -h
````

Each program has its own arguments, and the syntax will not necessarily be the same between programs. Often (but not always!) programs will reserve `-h` or `--help` to display a useage statement, which will usually give a list of common arguments and an example of how to run the program. Some programs also have a manual page that displays all useage options, which you can check at the command line with `man [name of command]`. Often times though it is just easiest to check the manual page by looking it up online!

### Searching for files
While `ls` is useful if we already know the path of a file or directory, often times we will want to be able to search recursively through a filesystem and find specific file(s) or directories; to do this, we can use the powerful `find` command. The syntax for `find` is slightly different from the commands we've looked at so far:

```
find [path] [arguments] [expression]
```

The `[path]` is the directory to start searching in; `find` will start there then recursively go thru down all subdirectories, sub-subdirectories, etc. to look for the target. One common piece of command line shorthand is the `./` expression, which specifies the *current directory you are in*, so `find ./` would start the search in your current directory.

As with other programs, we can modify `find`'s behavior using `[arguments]`, such as setting a maximum recursion "depth" for to search down the file tree (e.g. `-maxdepth 2` if we only wanted to search 2 levels "down").

The `[expression]` is the actual pattern that we are searching for, and can accept several different values. For instance, we can search for a specifically named file using the `-name` option:


````python
%%bash
find ./ -name 'example.fasta'
````

This is a good opportunity to introduce the concept of *wildcards*, denoted by the `*` symbol. This is a special "placeholder" that refers to *any number of any characters*. So, if we wanted to find every file that ends with the extension `.fasta` from our current directory on down through the file tree:

````python
%%bash
find ./ -name '*.fasta'
````

Wildcards are super useful, and can be used with many other command line tools, not just `find`!

### Visualizing files
There are numerous tools that are a part of most Unix-like systems that we can use to display the contents of a file:

-   `cat`: prints entire file to screen (short for con`cat`enate)
-   `head`: prints the first 10 lines of a file by default. We can change how many lines printed by adding the `-n #` argument, e.g. `head -n 5` prints only the first five lines
-   `tail`: the opposite of `head`, it prints the *last* 10 lines of a file
-   `less`: rather than printing the entire file contents to screen at once, `less` allows you to scroll through using the arrow keys (close `less` by pressing 'q'). In our specific workshop environment, `less` does not work when run within a code block, but at the command line it is very helpful when working with large files

`wc -l`: technically this is not a way to view a file's contents, but is still good to know! `wc` stands for 'word count', which as the name suggests counts the number of words in the file, and the `-l` argument instead makes it count the number of *lines* in the file.


### Input and output
Unix commands have three default data "streams": standard input (`stdin`), which passes data into a program (such as `ls`, `pwd`, etc.) usually by the keyboard. Standard output (`stdout`) is where the output of a command is sent, default usually being the screen. The final stream standard error (`stderr`) seperates output specifically for error messages, though by default it will also output to the screen.

We will often want to redirect these streams, most frequently the output, to other programs or to an output file. To do this we use the output overwrite character `>` after our command, followed by the name of a file; if the file does not exist, it will be created with its contents being the output of the command, while if the file does exist its contents will be OVERWRITTEN. For example, the following command will direct the output of the `ls` command to a text file:

```
ls ./ > directory_contents.txt
```

If we instead wanted to instead append to a file without overwriting its contents, we can use the `>>` operator...note that if the file does not exist it will still be created:

```
ls -l ./ >> directory_contents.txt
```

#### Using pipes
Rather than directing output to a file, we can also direct it *straight to another command*, which will use the output from the previous command as input. This is referred to as "piping" the output as it uses the **pipe operator** `|`, and is central to creating bioinformatic pipelines! For example, imagine we wanted to count the number of files in a directory. We could output a list from `ls` to a file, then count the number of lines in the file:

```
ls ./ > directory_contents.txt
wc -l directory_contents.txt
```

But that is messy, as it creates a needless intermediary file and involves extra typing! Instead, it is much more efficient to **pipe** one command into another:

```
ls ./ | wc -l
```

As we will see, pipes are an essetial part of our data wrangling toolkit, as they allow us to stitch together an arbitrary number of commands to manipulate and mutate files in only a single line of code!


## Intro to FASTA format
The first bioinformatic file type we'll be discussing is FASTA files, which is one of the most common formats used to represent biological sequences. FASTA as a format was originally taken from a bioinformatic software package written back in the 1980s, and stuck around and became a standard. A FASTA file is composed of a number of *entries*, which can be either protein or nucleotide sequences. An "entry" can represent a number of things, e.g.:

- Chromosomes in a genome assembly
- mRNA transcript sequences
- Translated protein sequences
- Etc.

Importantly, while the content of FASTA files can vary, the syntax is the same. Each entry is comprised of two parts:
- *Header line*, which starts with a `>` symbol. This contains the unique name for each entry (e.g. >chr1), optionally followed by additional data.
- *Sequence*, which follows a header line. This is a string of either nucleotides or amino acids that corresponds to the header

Apart from the unique identifier, the exact info in the header can vary based on the source of the file but must be only a single line. For the sequence, permitted characters are either single letter amino acid codes if a protein sequence, or the standard IUPAC nucleotide codes if nucelotide sequence (typically "ATCG" but also ambiguous nucleotide codes as well). Ambiguous nucleotides should be denoted by "N"s. Letters are typically uppercase but can also be lower case, which often represents "masked" sequence such as repetitive elements.

> Run the following code to see an example of what a FASTA file looks like:

```
less data_day1/example.fasta
```

The order is important, as the sequence must follow its corresponding header. However, for some FASTA files, the sequence line can be set to wrap every certain number of characters (e.g. every 50 nucleotides). These are referred to as *multi-line FASTA*, and can make your life difficult...make sure you know whether your FASTA is multi-line or single-line! For example:

```
>sequence1
ATGGACGCTAGTCAGTAGATGCATGCTGACCCAACATAACG
```

vs.

```
>sequence1
ATGGACGCTAG
TCAGTAGATGC
ATGCTGACCCA
ACATAACG
```

The sequences in these two examples are identical, but one is broken up by lines. Most programs can automatically handle both types of **FASTA** files, but as we stated above, if you parse your own **FASTA** files, you need to be aware of the difference!

#### Regular expressions

One last way that we can modify how `grep` behaves is with *regular expressions*, a.k.a *regex*. Regex are patterns that describe a *set of strings*. In other words, they allow you to match complex patterns with `grep` (and other Unix commands), not just exact matches! Regex syntax can be very convoluted, and if you are trying to match a complex pattern it is easiest to ask a LLM or the internet for the correct regex. However, there are a few basic regex patterns that are worth remembering, as they are useful in this sort of quick data wrangling:

-   `^` matches pattern at start of string
-   `$` matches pattern at end of string
-   `[ ]` matches any of enclosed characters. Can use in conjunction with 'A-Z' or '0-9', i.e.:
    -   `[A-Z|a-z]` matches any alphabetical character, upper or lower case
    -   `[0-9]` matches any numeric character

### Editing strings with `sed`
The Unix utility `sed` (`s`tream `ed`itor) is one such tool that allows us to perform text transformations, either on files or as input from a pipe. It reads input line by line and performs some function on each line. A `sed` command is structured like this:

```
sed [ARGUMENTS] '[COMMAND]' [INPUT]
```

If an input file is not specified, `sed` assumes the input is piped from standard input (aka `STDIN`). One of the most useful commands is the substitute function (`s/`), which replaces the first matching string in a line with another string:

```
sed 's/string_to_replace/string_replacing/' [INPUT]
```

For example:

````python
%%bash
#echo is another command line program that can be used to print text to the terminal
echo "oh time time thy pyramids" | sed 's/time/test/'
````

Note that by default, `sed` only replaces the first instance of a string in a line...if we want it to replace ALL matching strings in a line, we append `g` (for `g`lobal) to the command:

````python
%%bash
echo "oh time time thy pyramids" | sed 's/time/test/g'
````

#### Other `sed` functions
While string replacement is probably the most common use of `sed`, it has lots of other functionality that lets us quickly pull useful info out of files, such as using `sed -n` to print *specific numbered lines of a file*. For example, if we wanted to print just the 6th line of a file:

````python
%%bash
sed -n '6p' data_day1/assembly_1.fasta

#the -n tells `sed` to not automatically print each line unless explicitly told to
#the 'p' command tells `sed` to print the line if it matches the pattern (in this case, line 6)
````

We can specify a range of lines to print, such as the first command which will print lines 1 thru 5 of the file, or multiple non-sequential lines using the `-e` argument to add multiple expressions, as in the second command which prints only lines 1 and 5:

````python
%%bash
sed -n '1,5p' data_day1/assembly_1.fasta
````

````python
%%bash
sed -n -e '1p' -e '5p' data_day1/assembly_1.fasta
````

Another useful thing we can do with `sed` is **delete** specific lines of a file! The syntax is similar to printing (though without the `-n` argument), as shown in this command which will delete the 1st line from the file:

````python
%%bash
sed '1d' data_day1/assembly_1.fasta | head

#the 'd' command tells `sed` to delete the line if it matches the pattern (in this case, line 1)
````

And here, which will delete a range of lines from line 2 to 20,006:

````python
%%bash
sed '2,20006d' data_day1/assembly_1.fasta | head
````

In the above examples we referred to specific numbered lines as the pattern to match, but we can also provide a *specific string pattern* (similar to how we were substituting with `s/`) and delete any lines matching that pattern:

`sed '/pattern to match/d' [INPUT]`

So the following command would delete every line that matches a '>' character in our FASTA file:

````python
%%bash
sed '/>/d' data_day1/assembly_1.fasta | head -n 5
````

Something to note: if you remember back to the `grep` section, you may realize that the above `sed` command is basically equivalent to running inverted `grep` (i.e. `grep -v '>'`), which returns all lines that do NOT contain a specified pattern. This is a fairly common situation, where different tools can fulfill similar functions and there are several different viable ways to accomplish your task (similar to how `grep | wc -l` and `grep -c` do the same thing). So, which do you use? In most cases, it is up to you! There might be some differences in efficiency (which could start to matter if your files are very large), or some considerations depending on what exactly you are trying to do.

For example, if I am trying to directly modify a file in-place, I personally would prefer to use `sed` as it can be done in a single command using `-i`. I could use `grep` to direct output to a temporary file, then re-name that file:

```
grep -v '>' data_day1/assembly_1.fasta > temp
mv temp data_day1/assembly_1.fasta
```

But that is slightly more annoying to do, as it take a couple extra button presses!

## Seqkit
So far, we have been discussing generic text wrangler command line utilities that are part of most Unix systems and are useful for manipulating FASTA files. However, while they are powerful they were not designed specifically with FASTA or other sequence files in mind, so certain tasks might be difficult to accomplish using generic tools alone. Let's introduce our first specialized program `seqkit`, which is a [suite of tools](https://bioinf.shenwei.me/seqkit/) that were specifically designed to manipulate sequence files! We'll see that many of the tasks we have been using command line utilities for can also be done using `seqkit`, plus some extended functionality as well.

The downside to using `seqkit` vs generic tools is that it is not included by default as part of a Unix operating system like the other tools are, and must be downloaded manually. For this workshop we've already done that for you, but if you want to use it on your own you'll have to install it yourself. There are also plenty of alternative tools that have been developed as well, but we are focusing on `seqkit` as it is very functional, lightweight, and is actively maintained!

If we look at the [manual page](https://bioinf.shenwei.me/seqkit/usage/) for `seqkit`, we can see that it has a ton of different functions, with documentation and example useage for each. We aren't interested in all of these functions, so let's focus on a few common tasks.

### Manipulating and summarizing FASTA files
One very useful way to sanity check our FASTA files is checking sequence lengths; we can use `seqkit stats` to get a quick summary of length distribution, as well as some other basic summary statistics:

````python
%%bash
seqkit stats data_day1/assembly_1.fasta
````

`seqkit seq` can transform sequences in a variety of ways depending on the arguments used, such as:

````python
%%bash
#Convert lowercase to uppercase
seqkit seq -u data_day1/assembly_1.fasta | head
````

````python
%%bash
#Reverse complement sequence
seqkit seq -r -p data_day1/assembly_1.fasta | head
````

````python
%%bash
#Filter sequences shorter than a certain length (e.g., 100 bp)
seqkit seq -m 100 data_day1/assembly_1.fasta | head
````

Like the generic command line tools, we can use pipes to direct the input and output of `seqkit` commands into each other, allowing for some more advanced processing. For example, as detailed in the manual page for `seqkit stats`, the `stats` module will count any gaps that are present in the sequence, so if we wanted to get a stats summary of the file that does NOT include any gaps, we could use `seqkit seq -g` to remove gaps, then pipe to `seqkit stats` to get our summary:

````python
%%bash
seqkit seq -g data_day1/assembly_1.fasta | seqkit stats
````

Let's also look at `seqkit grep`, which has a familiar name! As the name suggests, it is inspired by command line `grep`, performing a similar function and shares a similar syntax. However, as `seqkit` was designed with FASTA files in mind, it has some additional handy functionality compared to normal `grep`. For example, we can search for a particular ID in the header, and it will return both the matching header AND the associated sequence:


````python
%%bash
#-p specifies the pattern to match with the sequence IDs in the headers of the FASTA file
seqkit grep -p 'chr1' data_day1/assembly_1.fasta | head
````

Another way that `seqkit grep` differs from standard `grep` is that by default it assume **exact pattern matching** (i.e. the complete opposite of command line `grep`!). If we wanted to instead for e.g. match any sequence ID that contains the string 'chr', we can use the argument `-r` to enable partial matches using regular expressions (regex):

````python
%%bash
#-p specifies the pattern to match with the sequence IDs in the headers of the FASTA file
#-r enables partial pattern matching, so this will match any sequence that has 'chr' in the header
seqkit grep -r -p 'chr' data_day1/assembly_1.fasta > data_day1/assembly_1_chr_sequences.fasta

grep '^>' data_day1/assembly_1_chr_sequences.fasta
````

### Editing FASTA files
We discussed some basic editting using `sed`, but `seqkit replace` also fills a similar function, and allows for some more advanced editting. In basic useage, provide a pattern to match with `-p` and a replacement for the pattern with `-r`, e.g. if we wanted to change 'chr' to 'chromosome':

````python
%%bash
seqkit replace -p 'chr' -r 'chromosome' data_day1/assembly_1.fasta | grep '^>'
````

This is something you could also do with `sed`, however we can see in the example useage section of the documentation that we can see that we can make edits that would be much trickier to do if you were using just `sed`. For example, if we wanted to add a prefix string to each entry in the FASTA file, we can use the regex for "start of" (`^`) to prepend the string:

````python
%%bash
seqkit replace -p ^ -r sequence_ data_day1/assembly_1.fasta | grep '^>'
````

## FASTQ files
The other type of sequence file format we will discuss is FASTQ, which as the name suggests, is similar to FASTA format but also contains **q**uality information about the sequence. You will commonly encounter FASTQ files when working with sequencing pipelines, as second generation technologies like Illumina as well as 3rd generation technologies like PacBio and Nanopore output their data in this format. We aren't going to spend as much time on FASTQ compared to FASTA, as in the context of bioinformatic analysis, while FASTQ files are often the starting point in a pipeline you likely won't be doing much manipulation of the files themselves. However, it is a very common format, so it is still important to be familiar with them.

Each entry is comprised of *four* lines, as opposed to the two of FASTA:

-   *Header line* which starts with an `@` symbol and contains the sequence ID
-   *Sequence line* comprised of nucleotides
-   *Spacer line* which is just a `+` character (an optionally the sequence ID again)
-   *Quality line* which is a string of ASCII characters, each character corresponding to a base in the nucleotide sequence

Let's have a look at an example:

````python
%%bash
head -n 8 data_day1/example.fastq
````

Here we see two sequences, `@SRR1234567.116808442/1` and `@SRR1234567.116808442/2`. The line directly following the header contains the familiar ATCG nucleotide symbols, and then there is a line with a `+`, simply meant to provide some space. Next, we see a line with seemingly random characters.

These characters correspond to the *Phred quality score* of the base at the same position in the sequence string. So the first quality score of the first sequence, `7`, matches with the first base in the sequence string, `T`, and so on. The **quality score** is actually a numerical value encoded as a single character and is expressed by a *Q score*. These reflecte the probability that that base is accurate. Q scores are calculated by:

Q = -10 \* log10(P)

where P = the error calling probability for that base. In short, *a higher Q score == a higher confidence call*, e.g. Q10 = 90% base call accuracy, Q30 = 99.9% accuracy, etc. To get its corresponding symbol, add +33 to the Q score to get the ASCII code and take the symbol that corresponds to that code; e.g. Q10 == `+`, Q30 == `?`. For reference, [here](https://www.omixon.com/wp-content/uploads/2013/06/illumina_fastq_coding.png) is a table that lists Q scores and their corresponding ASCII codes and values. Most programs that handle FASTQ files translate these symbols for you, so don't worry about having to convert back and forth manually!

### Paired FASTQ files
As mentioned, FASTQ files are commonly seen when performing sequencing experiments, either of transcripts or whole genome. When dealing with short read Illumina sequencing (e.g. in RNAseq or population resequencing experiments) you will most often have a *pair* of FASTQ files. This is because Illumina uses "paired-end" sequencing libraries, which basically means that *each sequencing read will have a linked partner read in the other file*, which is reflected in the sequence ID. For example, in the file that we looked at above we can see that both entries share the same `SRR1234567.116808442` ID, followed by a `1` or a `2`, indicating that they are paired. We will discuss how this paired information can be used when aligning reads to a reference in the next workshop when we discuss alignment files!
